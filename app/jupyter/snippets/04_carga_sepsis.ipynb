# Carga y exploración inicial — Sepsis (PhysioNet Challenge 2019)

El dataset de sepsis está organizado en ficheros `.psv`, con información temporal de pacientes de UCI.

En este notebook veremos primero un único paciente y después cargaremos todos los pacientes del conjunto.

**Objetivos:**
- Cargar un fichero `.psv`.
- Inspeccionar sus dimensiones y columnas.
- Visualizar los primeros 20 registros.
- Cargar posteriormente todos los pacientes en un único DataFrame.

```python
import pandas as pd
import glob
```

```python
# Cargar un paciente de ejemplo
paciente = pd.read_csv(
    "/home/jovyan/datos/sepsis-2019/training/training_setA/p000001.psv",
    sep="|"
)
```

```python
# Información básica
print(f"Filas: {paciente.shape[0]}")
print(f"Columnas: {paciente.shape[1]}")
```

```python
# Nombres de las columnas
paciente.columns.tolist()
```

```python
# Primeros 20 registros del paciente
paciente.head(20)
```

```python
# Cargar todos los pacientes del conjunto training_setA
ficheros = glob.glob(
    "/home/jovyan/datos/sepsis-2019/training/training_setA/*.psv"
)

todos = pd.concat(
    (
        pd.read_csv(f, sep="|").assign(paciente=f)
        for f in ficheros
    ),
    ignore_index=True,
)

print(f"Pacientes/ficheros encontrados: {len(ficheros)}")
print(f"Filas totales: {todos.shape[0]}")
print(f"Columnas totales: {todos.shape[1]}")
```

```python
# Primeros 20 registros del conjunto completo
todos.head(20)
```

## Primera observación

Fíjate especialmente en que los datos de sepsis tienen una **dimensión temporal**:

- `HR`, `O2Sat`, `Temp`, etc. representan medidas a lo largo del tiempo.
- `SepsisLabel` indica la etiqueta asociada a cada instante.
- La variable `Time` permite interpretar la evolución temporal.
- Cada paciente aparece en varias filas.
