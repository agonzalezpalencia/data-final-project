# Carga y exploración inicial — Arritmia (MIT-BIH)

En este notebook cargaremos un registro de ECG del dataset MIT-BIH Arrhythmia.

A diferencia de los otros datasets, aquí trabajamos directamente con una **señal** y sus anotaciones.

**Objetivos:**
- Cargar un registro ECG.
- Obtener la señal de los canales disponibles.
- Cargar las anotaciones de los latidos.
- Visualizar información básica.
- Representar una pequeña ventana de la señal.


```python
# Instalar las librerías necesarias
!pip install wfdb matplotlib
```

```python
import wfdb
import matplotlib.pyplot as plt
```

```python
# Cargar un registro de ejemplo
registro = wfdb.rdrecord(
    "/home/jovyan/datos/mit-bih-arrhythmia-1.0.0/100"
)

# Cargar las anotaciones de los latidos
anotaciones = wfdb.rdann(
    "/home/jovyan/datos/mit-bih-arrhythmia-1.0.0/100",
    "atr"
)

senal = registro.p_signal
posiciones = anotaciones.sample
etiquetas = anotaciones.symbol
```

```python
# Información básica del registro
print(f"Frecuencia de muestreo: {registro.fs} Hz")
print(f"Número de muestras: {senal.shape[0]}")
print(f"Número de canales: {senal.shape[1]}")
print(f"Número de anotaciones: {len(anotaciones.sample)}")
```

```python
# Primeras 20 anotaciones
for posicion, etiqueta in zip(posiciones[:20], etiquetas[:20]):
    print(f"Muestra: {posicion:>6} | Etiqueta: {etiqueta}")
```

```python
# Visualizar una pequeña ventana de la señal
# Se muestran las primeras 2000 muestras del primer canal.

plt.figure(figsize=(14, 4))
plt.plot(senal[:2000, 0])
plt.title("Registro ECG 100 — primeras 2000 muestras")
plt.xlabel("Muestra")
plt.ylabel("Amplitud")
plt.show()
```

## Primera observación

Observa la relación entre:

- la **señal ECG**;
- las **posiciones** de los latidos;
- las **etiquetas** asociadas a esos latidos.
